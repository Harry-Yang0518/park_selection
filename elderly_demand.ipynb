```python
import pandas as pd
import os
```

```python
input_file = './dataset_structured/02_vector_tables/elderly_population/attributes.csv'
output_dir = './dataset_structured/05_processed'
```

```python
pop_df = pd.read_csv(input_file)
pop_df['elderly_count'] = pop_df['population'] * pop_df['elder_rati']
```

```python
def normalize(col):
    return (col - col.min()) / (col.max() - col.min())

pop_df['demand_index'] = normalize(pop_df['elderly_count'])
```

```python
output_file = f'{output_dir}/processed_elderly_demand.csv'
pop_df.to_csv(output_file, index=False)

print(f"--- SUCCESS: Demand calculated ---")
print(f"Results saved to: {output_file}")
print(f"Top 5 high demand areas (abs count):\n", pop_df[['id', 'elderly_count']].sort_values(by='elderly_count', ascending=False).head())
```
