```python
import geopandas as gpd
import os
```

```python
elder_pop_dir = r'C:\laura\NYUSH\Course\8_2026Spring\urban\final_project\公园数据\F\4\Elde pop(_7%with index)-20260318T034124Z-1-001\Elde pop(_7_with index)'
elder_pop_shp = os.path.join(elder_pop_dir, 'filtered_elder_pop1106_1.shp')

output_dir = r'C:\laura\NYUSH\Course\8_2026Spring\urban\park_selection\dataset_structured\05_processed\standardized_layers'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
print("Loading elderly population shapefile...")
gdf_demand = gpd.read_file(elder_pop_shp)

gdf_demand = gdf_demand.to_crs(epsg=3857)

gdf_demand['elderly_demand'] = gdf_demand['population'] * gdf_demand['elder_rati']
```

```python
def normalize(col):
    return (col - col.min()) / (col.max() - col.min())

gdf_demand['demand_index'] = normalize(gdf_demand['elderly_demand'])
```

```python
save_path = os.path.join(output_dir, 'elderly_demand_3857.shp')
gdf_demand.to_file(save_path, encoding='utf-8')

print(f"Success! Calculated elderly demand for {len(gdf_demand)} grids.")
print(f"Saved to: {save_path}")

print(gdf_demand.head())
```
