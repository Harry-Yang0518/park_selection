```python
import geopandas as gpd
import pandas as pd
```

```python
gdf_gap = gpd.read_file(r'./dataset_structured/05_processed/standardized_layers/accessibility_final_grid.shp')
```

```python
# normalize
gdf_gap['norm_A'] = (gdf_gap['Ai_access'] - gdf_gap['Ai_access'].min()) / (gdf_gap['Ai_access'].max() - gdf_gap['Ai_access'].min())
gdf_gap['norm_D'] = (gdf_gap['elderly_de'] - gdf_gap['elderly_de'].min()) / (gdf_gap['elderly_de'].max() - gdf_gap['elderly_de'].min())
```

```python
# Gap = norm_D * (1 - norm_A)
gdf_gap['gap_score'] = gdf_gap['norm_D'] * (1 - gdf_gap['norm_A'])

# choose top 5% as potential new park spots
potential_spots = gdf_gap.sort_values(by='gap_score', ascending=False).head(int(len(gdf_gap)*0.01))

print(f"识别出 {len(potential_spots)} 个高优先级缺口格网。")
```

```python
# gdf_gap.to_file(r'./dataset_structured/05_processed/standardized_layers/gap_analysis.shp')
potential_spots.to_file(r'./dataset_structured/05_processed/standardized_layers/candidate_new_park_spots.shp')
```
