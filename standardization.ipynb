```python
print("Transforming POI Data...")
df_poi = pd.read_csv(input_poi_csv)
gdf_poi = gpd.GeoDataFrame(
    df_poi,
    geometry=gpd.points_from_xy(df_poi.lon_wgs, df_poi.lat_wgs),
    crs="EPSG:4326"
)
gdf_poi_3857 = gdf_poi.to_crs("EPSG:3857")
gdf_poi_3857.to_file(os.path.join(output_dir, 'pois_3857.shp'), encoding='utf-8')
```
